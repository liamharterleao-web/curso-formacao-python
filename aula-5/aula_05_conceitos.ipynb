## 1. Tabela verdade

Cada condição é `True` (V) ou `False` (F). A tabela lista **todas** as combinações.

| A | B | A and B | A or B |
|---|---|---------|--------|
| V | V | V | V |
| V | F | F | V |
| F | V | F | V |
| F | F | F | F |

| A | not A |
|---|-------|
| V | F |
| F | V |

Rode a célula abaixo e confira se bate com a tabela.


```python
print("True and True  =", True and True)
print("True and False =", True and False)
print("False and True =", False and True)
print("False and False=", False and False)
print()
print("True or True   =", True or True)
print("True or False  =", True or False)
print("False or True  =", False or True)
print("False or False =", False or False)
print()
print("not True  =", not True)
print("not False =", not False)
```

## 5. `not` — não tem cupom

String vazia é tratada como falsa. Rode. Depois coloque `LANCHE10`.


```python
cupom = ""
if not cupom:
    print("sem cupom")
else:
    print("cupom informado")

```

## 6. Embalagem: viagem **e** total baixo

Complete. Viagem `s` e valor `8.50` → soma 2.00. Viagem `s` e `30.60` → não soma.


```python
viagem = "s"
valor = 8.50

if viagem == "s" and valor < 15:
    valor = valor + 2.00
    print("embalagem cobrada")
print(f"R$ {valor:.2f}")

```

## 7. Junte as regras (números fixos)

Quantidade 4, preço 8.50, cupom `LANCHE10`, viagem `n` → `R$ 30.60`.
Depois teste quantidade `0` (recusa, sem total).


```python
import sys


quantidade = 4
preco = 8.50
cupom = "LANCHE10"
viagem = "n"

# complete: recusa, subtotal, cupom com and, embalagem, print

if quantidade == 0:
    sys.exit()

if quantidade == 4 and preco == 8.50 and cupom == "LANCHE10" and viagem == "n":
    subtotal = quantidade * preco
    print(f"Subtotal: R$ {subtotal:.2f}")
```

## 8. `match` / `case` — o switch do Python

Quando **um** valor escolhe entre opções nomeadas, use `match`. `case _` = qualquer outra coisa.

Rode com `forma = "dinheiro"`. Depois troque para `"boleto"`.


```python
valor = 17.0
forma = "dinheiro"  # troque para "cartao", "pix" ou "boleto"

match forma:
    case "dinheiro":
        total = valor * 0.9
    case "cartao":
        total = valor * 1.05
    case "pix":
        total = valor
    case _:
        print("Forma desconhecida.")
        total = None

if total is not None:
    print(f"R$ {total:.2f}")
```

## 9. Quando usar o quê

| Situação | Preferir |
|----------|----------|
| Faixa (`quantidade < 1`) | `if` |
| Duas condições (cupom) | `and` / `or` |
| Opções nomeadas (forma) | `match` |

`match` **não** substitui `and` no cupom.
